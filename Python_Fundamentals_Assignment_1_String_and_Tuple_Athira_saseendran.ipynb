## Strings (Concatenation, Slicing and Other methods) :
 ### 1. String Concatenation
Write a Python program that takes two strings i.e string 1 “Hello ”, string 2 get name as
input from the user and concatenates them together. Display the concatenated string as
the output.
Then concatenate string 3 “, welcome to Python programming” to the existing string and
display the output string

```python
# Get the name from the user
name = input("Enter your Name: ")

# Store the strings in variables
string1 = "Hello "
string2 = name

# Concatenate(combine) the strings
result = string1 + string2
print(result)
string3 = ", welcome to Python programming"
result = result + string3
print(result)
```

I used input() function to get the user's name. I used the "+" operator to concatenate the strings.First "Hello" is combined with the name.Then I added "welcome to Python programming" to the existing string.

 ###  2. String Slicing and Indexing:
 Write a Python program using the above concatenated string as input and performs the
following tasks:

a. Print the first character of the string.

b. Print the last character of the string.

c. Print the first 5 characters of the string.

d. Print the last 11 characters of the string

e. Print the string in reverse.

f. Use slicing and print the word “Python” from the existing string.

```python
# Print the first character
print("First character:" , result[0])
```

```python
# Print the last character
print("Last character:",result[-1])
```

```python
# Print the first 5 characters
print("First 5 characters:",result[:5])
```

```python
# Print the last 11 characters
print("Last 11 characters:",result[-11:])
```

```python
# Print the string in reverse
print("Reverse:",result[::-1])
```

I used "Indexing" to access individual characters and "Slicing" to get the parts of the string.  I used [::-1] to reverse the string.

 ### 3. String Methods
Write a Python program that takes a string, strM = “Python beginner tutorial” and perform the following tasks:

a. Convert the sentence to uppercase.

b. Convert the sentence to lowercase.

c. Use Capitalize and return the sentence to the original input form.

d. Count the total number of occurrences of character ‘t’ in the string.

e. Replace all occurrences of “Python” with “Machine Learning” in the input string
strM = “Python beginner tutorial”


```python
strM = "Python beginner tutorial"
```

```python
# Convert the sentence to uppercase
print("Uppercase:",strM.upper())
```

```python
# Convert the sentence to lowercase
print("Lowercase:",strM.lower())
```

```python
# Capitalize the sentence
print("Capitalize:", strM.capitalize())
```

```python
# Count the number of times "t" occurs
print("Count of 't':", strM.count("t"))
```

```python
# Replace Python with Machine Learning
print("After replacement:", strM.replace("Python", "Machine Learning"))
```

I used built in string methods to perform different operations. upper() converts the string to uppercase, lower() converts it to lowercase, capitalize() capitalizes the first letter, count() counts the occurrences of a character (here t), and replace() replaces one part of the string with another.

##Tuples (Creation, Modification and Access)
Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

a. Concatenate the two tuples and store it in “t_combine”

b. Repeat the elements of “t_combine” 3 times

c. Access the 3rd element from “t_combine”

d. Access the first three elements from “t_combine”

e. Access the last three elements from “t_combine”

```python
# Create the two tuples
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)
```

```python
# Concatenate the two tuples
t_combine = tuple1 + tuple2
print("Combined tuple:", t_combine)
```

```python
# Repeat the elements 3 times
print("Repeated tuple:", t_combine * 3)
```

```python
# Access the 3rd element
print("3rd element:", t_combine[2])
```

```python
# Access the first three elements
print("First three elements:", t_combine[:3])
```

```python
# Access the last three elements
print("Last three elements:", t_combine[-3:])
```

I created two tuples and used the "+" operator to combine them. I used "*" operator to repeat the tuple three times. I used "Indexing' to access the third element and "Slicing" to access the first and last three elements.
